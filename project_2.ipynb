```python
#import fils from sys 
import sys
#specific the data folder  
sys.path.append(r"M:\DPI COURSE\tech py\project 2")
#libirary to count inside lists 
from collections import Counter

#import data from countires fil
from countries import countries

from countries_details_dat import countries_data
countries_with_land = []
for country in countries:  # 2. حلقة تدور على كل دولة
    if "land" in country.lower():  # 3. شرط: هل كلمة land موجودة في اسم الدولة؟
        countries_with_land.append(country)
print("____1-____the country with (land)___")
for index, country in enumerate(countries_with_land, 1):
    print(f"{index} - {country}")



# import fata from the file 
total_language = set()
for languages in countries_data:
    for n_lang in languages["languages"]:
     total_language.add(n_lang) 
print(f"___2___-total language in file is : {len(total_language)}")

for index, languages in enumerate(total_language,1):
     print(f" {index}-{languages}")

#most 3 spoken language 
spoken_language = []
for s_language in countries_data:
    spoken_language.extend(s_language["languages"])
#counter
count_language = Counter(spoken_language)  
most_3_spoken = count_language.most_common(3)
print(f"____3____-\nmost_3_spoken is :\n {most_3_spoken}") 




```
